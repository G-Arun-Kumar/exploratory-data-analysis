**D tale is an amazing library used for EDA. complete EDA can be done in single line of code.**

```python
import dtale
import seaborn as sns
%matplotlib inline
```

```python
# Remember, we can load the predefined datasets using seaborn.
# We need to just give name of predefined dataset and not the complete path.

# If we need to load our dataset, go with the obvious way of pandas functions pd.read_csv().
df = sns.load_dataset('titanic')
dtale.show(df)
```

## pandas-profiling/ydata-profiling

**The next automated library for EDA is pandas-profiling**.
This is a very good automated library which involve only one line of code that comprises all information of EDA in your datasets.

Remember, to install, we need to give the name of the module as **pandas-profiling**. But while we import, we need to give the name as **pandas_profiling**

Remember, this works well only when python version is less than 3.11+. For version above 3.11, install & use **ydata-profiling**

ydata_profiling is exactly the same as pandas-profiling.

```python
df1 = sns.load_dataset('tips')
from ydata_profiling import ProfileReport
profile = ProfileReport(df1, explorative=True)
profile.to_file(r'E:\Study\Projects\EDA\output_tips_data.html')
```
