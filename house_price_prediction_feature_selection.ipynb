```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Let's import the required data
df = pd.read_csv(r'E:\Study\Projects\EDA\output_feature_engineerig_data.csv')
pd.set_option('display.max_columns', None)
```

```python
# Since we have 0.2% of missing values of overall dataframe, we will drop it
#null_count = df.isnull().sum().sum()
#total_values = df.size
#percentage_missing = (null_count/total_values) * 100
#print(f"The total percent of missing values is {percentage_missing}")
df.dropna(inplace=True)
```

```python
df.head()
```

```python
# We will now split the data into X_train and y_train to build the model.
# y_train will be the dependent variable vector(SalePrice)
y_train = df['SalePrice']
```

```python
# Now since we don't need 'Id' and 'SalePrice' in X_train varaible, we will drop those 2 columns
X_train = df.drop(['Id', 'SalePrice'], axis=1)
```

```python
# Now let's import the LASSO & SelectFromModel class to select the best required features for building model
from sklearn.linear_model import Lasso
from sklearn.feature_selection import SelectFromModel
feature_selection = SelectFromModel(Lasso(alpha=0.01, random_state=11))
feature_selection.fit(X_train, y_train)
```

```python
feature_selection.get_support()
```

```python
selected_features = X_train.columns[(feature_selection.get_support())]

print(f"The overall count of feature is {X_train.shape[1]}")
print(f"The unwanted features that are removed is {len(selected_features)}")
print(f"The no. of features that have zero coefficients {np.sum(feature_selection.estimator_.coef_ == 0)}")
```
